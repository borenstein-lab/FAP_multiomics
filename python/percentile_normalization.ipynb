```python
import pandas as pd
import os
```

#### Data preparation

```python
project_dir = "C:/Users/alisa/phd/FAP"
work_dir = os.path.join(project_dir,"qiime/batch_correct")
mdata = pd.read_csv(os.path.join(work_dir, "core_metadata_cross_cohort.csv"))
```

```python
# Create a dictionary with sample IDs for each cohort and case/control status
sample_dict = {
    "control_internal": "\t".join(mdata[(mdata.phenotype_binary == "Control") & (mdata.cohort == "Internal")]['sample_id'].astype(str).str.strip().tolist()),
    "fap_internal": "\t".join(mdata[(mdata.phenotype_binary == "FAP") & (mdata.cohort == "Internal")]['sample_id'].astype(str).str.strip().tolist()),
    "control_katona": "\t".join(mdata[(mdata.phenotype_binary == "Control") & (mdata.cohort == "Katona")]['sample_id'].astype(str).str.strip().tolist()),
    "fap_katona": "\t".join(mdata[(mdata.phenotype_binary == "FAP") & (mdata.cohort == "Katona")]['sample_id'].astype(str).str.strip().tolist())
}
```

```python
# Save each cohort's sample IDs to a separate text file
for var_name, content in sample_dict.items():
    with open(f"{work_dir}/{var_name}.txt", "w") as f:
        f.write(content)
```

#### Percentile normalization

```python
# Calculate percentile normalization for the internal cohort
input_file_path = os.path.join(project_dir, "processed_data/genus_ra_non_rare.csv")
case_list_path = os.path.join(work_dir, "fap_internal.txt")
control_list_path = os.path.join(work_dir, "control_internal.txt")
output_file_path = os.path.join(work_dir, "genus_ra_non_rare_perc_norm.tsv")

command = (f"python percentile_norm.py -i {input_file_path} -case {case_list_path} -control {control_list_path} -o {output_file_path} -otu-d comma")
os.system(command)
```

```python
# Calculate percentile normalization for the Katona cohort
input_file_path = os.path.join(project_dir, "processed_data/katona_non_rare.csv")
case_list_path = os.path.join(work_dir, "fap_katona.txt")
control_list_path = os.path.join(work_dir, "control_katona.txt")
output_file_path = os.path.join(work_dir, "katona_non_rare_perc_norm.tsv")

command = (f"python percentile_norm.py -i {input_file_path} -case {case_list_path} -control {control_list_path} -o {output_file_path} -otu-d comma")
os.system(command)
```
